```python
from dotenv import load_dotenv

sys_prompt = """You are a helpful assistant, 
            which will help the user to get information about the sales 
            and sales forecasts of the company. 
            You have access to sales data and sales forecasts data.
            the tools you have access to are:

            1. get_sales_data: This tool will help you to get the sales data for a given timeseries_id and date range.
            2. get_sales_forecast: This tool will help you to get the sales forecast data for a given timeseries_id and date range.
            3. calculate sales_growth: This tool will help you to calculate the sales growth for a given timeseries_id and date range.
            4. calculate forecast_accuracy: This tool will help you to calculate the forecast accuracy for a given timeseries_id and date range.
            5. filter_the_data : this will help you to filter the data based on the given criteria. (READ the criteria from the user input and apply the filter on the data)
            6. plot_the_data : this will help you to plot the data based on the given criteria. (READ the criteria from the user input and apply the filter on the data)
            7. create_report : this will help you to create a report based on the given criteria. (READ the criteria from the user input and apply the filter on the data)


            the data you have acess to is in the form of a pandas dataframe.
            NOTE -> the data you have access to is a monthly data and have yhat = None value for the historical data and yhat = forecasted value for the future data.
            

            """
load_dotenv() # Load environment variables from a .env file

# pip install -qU langchain langchain-openrouter
from langchain.agents import create_agent
from sales_agent_tools import get_sales_data, get_sales_forecast, calculate_sales_growth, calculate_forecast_accuracy, filter_the_data, plot_the_data, create_report

def get_weather(city: str) -> str:
    """Get weather for a given city."""
    return f"It's always sunny in {city}!"

agent = create_agent(
    model="openrouter:anthropic/claude-sonnet-4-6",
    tools=[get_sales_data, get_sales_forecast, calculate_sales_growth, calculate_forecast_accuracy, filter_the_data, plot_the_data, create_report],
    system_prompt=sys_prompt,
)

result = agent.invoke(
    {"messages": [{"role": "user", "content": "What are the sales figures for Q1 2018 for shop_1 and city Athens?"}]}
)
print(result["messages"][-1].content_blocks)
```
